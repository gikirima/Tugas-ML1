```python
from sklearn.datasets import load_iris
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# 1. Load dataset
iris = load_iris()
X = pd.DataFrame(iris.data, columns=iris.feature_names)
y = iris.target

print(X.describe())
```

```python
# 2. Pipeline: Preprocessing + PCA (untuk visualisasi)
pca_pipe = Pipeline([
    ('scaler', StandardScaler()),
    ('pca', PCA(n_components=2))
])
X_pca = pca_pipe.fit_transform(X)
```

```python
# 3. t-SNE (di luar pipeline karena tidak lazim dipakai di dalam)
X_scaled = StandardScaler().fit_transform(X)
X_tsne = TSNE(n_components=2, random_state=42).fit_transform(X_scaled)
```

```python
# 4. Visualisasi PCA vs t-SNE
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
sns.scatterplot(x=X_pca[:, 0], y=X_pca[:, 1], hue=y, palette="Set1", ax=axes[0])
axes[0].set_title("PCA (Label Asli)")
sns.scatterplot(x=X_tsne[:, 0], y=X_tsne[:, 1], hue=y, palette="Set1", ax=axes[1])
axes[1].set_title("t-SNE (Label Asli)")
plt.suptitle("Perbandingan PCA dan t-SNE")
plt.tight_layout()
plt.show()
```

```python
# 5. Pipeline: Preprocessing + Clustering
cluster_pipe = Pipeline([
    ('scaler', StandardScaler()),
    ('kmeans', KMeans(n_clusters=2, random_state=42))
])
cluster_pipe.fit(X)
labels = cluster_pipe.named_steps['kmeans'].labels_
```

```python
# 6. Evaluasi hasil clustering
sil_score = silhouette_score(X_scaled, labels)
inertia = cluster_pipe.named_steps['kmeans'].inertia_
print(f"Inertia (k=2): {inertia}")
print(f"Silhouette Score (k=2): {sil_score:.4f}")
```

```python
from kneed import KneeLocator

# 7. Elbow Method
inertias = []
K = range(1, 11)
for k in K:
    pipe = Pipeline([
        ('scaler', StandardScaler()),
        ('kmeans', KMeans(n_clusters=k, random_state=42))
    ])
    pipe.fit(X)
    inertias.append(pipe.named_steps['kmeans'].inertia_)

knee = KneeLocator(K, inertias, curve='convex', direction='decreasing')
```

```python
# Plot elbow
plt.figure(figsize=(8, 5))
plt.plot(K, inertias, 'bo-')
plt.vlines(knee.knee, min(inertias), max(inertias), linestyles='dashed', colors='red', label=f"Elbow at k={knee.knee}")
plt.xlabel("Jumlah Klaster (k)")
plt.ylabel("Inertia")
plt.title("Metode Elbow")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# pipeline dengan hyperparameter k=4
cluster_pipe_k4 = Pipeline([
    ('scaler', StandardScaler()),
    ('kmeans', KMeans(n_clusters=4, random_state=42))
])
cluster_pipe_k4.fit(X)
labels_k4 = cluster_pipe_k4.named_steps['kmeans'].labels_

X_pca_clustered = PCA(n_components=2).fit_transform(X_scaled)

# 8 & 9. Visualisasi Hasil Clustering dan Label Asli (Side-by-Side)
fig, axes = plt.subplots(1, 2, figsize=(12, 5))  # 1 baris, 2 kolom

# Hasil Clustering
sns.scatterplot(x=X_pca_clustered[:, 0], y=X_pca_clustered[:, 1], hue=labels_k4, palette='Set2', ax=axes[0])
axes[0].set_title("PCA - Hasil Clustering (k=4)")
axes[0].set_xlabel("PC1")
axes[0].set_ylabel("PC2")

# Label Asli
sns.scatterplot(x=X_pca_clustered[:, 0], y=X_pca_clustered[:, 1], hue=y, palette='Set1', ax=axes[1])
axes[1].set_title("PCA - Label Asli")
axes[1].set_xlabel("PC1")
axes[1].set_ylabel("PC2")

plt.tight_layout()  # Untuk mengatur jarak antar grafik
plt.show()
```
